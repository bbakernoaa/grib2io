# Xarray Backend Demo

This notebook demonstrates how to use the grib2io xarray backend to open and work with GRIB2 files as xarray datasets and datatrees.

## Import Required Libraries

```python
import grib2io
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
```

## Opening GRIB2 Files with Xarray Backend

### Basic 

Here we will open the grib2 file that has two types of surfaces.  First lets open the surface values or `typeOfFirstFixedSurface=1`. 

```python
# Open a GRIB2 file using the grib2io xarray backend with a filter for typeOfFirstFixedSurface
filters = {'typeOfFirstFixedSurface': 1}  # 1 = Ground or Water Surface
ds = xr.open_dataset('../tests/input_data/gefs.chem.t00z.a2d_0p25.f000.grib2_subset', engine='grib2io', filters=filters)
print(ds)
```

### Using Filters

```python
# Open with filters to select specific messages
filters = {
    'shortName': 'sfc_tot_pm25',
    'typeOfFirstFixedSurface': 1
}
ds_filtered = xr.open_dataset('../tests/input_data/gefs.chem.t00z.a2d_0p25.f000.grib2_subset',
                             engine='grib2io',
                             filters=filters)
print(ds_filtered)
```

### Using Data Model

```python
# Open with data model for standardized metadata
ds_nws_viz = xr.open_dataset('../tests/input_data/gfs.jpeg.grib2',
                           engine='grib2io',
                           data_model='nws-viz')
print(ds_nws_viz)
```

## Working with DataTree

### Opening as DataTree

```python
# Open a GRIB2 file as a DataTree
tree = xr.open_datatree('../tests/input_data/gefs.chem.t00z.a2d_0p25.f000.grib2_subset', engine='grib2io')
print(tree)
```

### DataTree with Stack Vertical

```python
# Open DataTree with vertical levels stacked
tree_stacked = xr.open_datatree('tests/input_data/gfs.jpeg.grib2',
                               engine='grib2io',
                               stack_vertical=True)
print(tree_stacked)
```

## Data Access and Manipulation

### Accessing Data

```python
# Access data from dataset
print('Variables in dataset:', list(ds.data_vars))
print('Coordinates:', list(ds.coords))

# Get a specific variable
temp_data = ds['TMP']
print(temp_data)
```

### Data Interpolation

```python
# Interpolate data to a new grid
grid_def_out = grib2io.Grib2GridDef(0, [0, 721, 1440, 90000000, 0, 128, 0, 0, 359000000, 90000000, 360000000, 0])
interp_ds = ds.grib2io.interp('bilinear', grid_def_out)
print(interp_ds)
```

### Subsetting Data

```python
# Subset data geographically
subset_ds = ds.grib2io.subset(lats=(24.5, 49.4), lons=(230, 295))
print(subset_ds)
```

## Writing Data Back to GRIB2

### Writing Dataset to GRIB2

```python
# Write dataset back to GRIB2 file
ds.grib2io.to_grib2('output.grib2', mode='w')
```

### Writing DataTree to GRIB2

```python
# Write DataTree back to GRIB2 file
tree.grib2io.to_grib2('output_tree.grib2', mode='w')
```

## Advanced Features

### Updating Attributes

```python
# Update attributes of a DataArray
da = ds['TMP']
updated_da = da.grib2io.update_attrs(shortName='TEMP', units='K')
print(updated_da)
```

### Working with Multiple Files

```python
# Open multiple files as a single dataset
ds_multi = xr.open_mfdataset([
    'tests/input_data/gfs.jpeg.grib2',
    'tests/input_data/gfs.png.grib2'
], engine='grib2io')
print(ds_multi)
```

## Visualization

### Plotting Data

```python
# Plot temperature data
temp_data = ds['TMP']
temp_data.isel(level=0).plot()
plt.title('Temperature at First Level')
plt.show()
```

### Plotting with Cartopy

```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Create a plot with cartopy
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())

# Plot data
temp_data.isel(level=0).plot(ax=ax, transform=ccrs.PlateCarree())

# Add map features
ax.coastlines()
ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.add_feature(cfeature.STATES, linestyle=':')

plt.title('Temperature Plot')
plt.show()
```

## Best Practices

### Memory Management

```python
# Close datasets when done to free memory
ds.close()
ds_filtered.close()
ds_nws_viz.close()
```

### Using Context Managers

```python
# Use context managers for automatic cleanup
with xr.open_dataset('tests/input_data/gfs.jpeg.grib2', engine='grib2io') as ds:
    # Work with data here
    print(ds)
    # Dataset automatically closed when exiting the block
```

```python
# Example: Stream GFS GRIB2 data from AWS S3 using s3fs, fsspec, and grib2io's xarray interface
import s3fs
import xarray as xr
import grib2io

# S3 path to a small GFS GRIB2 file
s3_url = "s3://noaa-gfs-bdp-pds/gfs.20250220/12/atmos/gfs.t12z.pgrb2b.1p00.f000"
# Create an s3fs filesystem object
fs = s3fs.S3FileSystem(anon=True)

# Open the file using s3fs and grib2io's xarray backend
# Note: For S3 data with multiple surface types, use specific filters
# Let's filter for isobaric surface (code 100) and a specific variable
filters = {
    'typeOfFirstFixedSurface': 100,  # Isobaric Surface
    'shortName': 'TMP'  # Temperature variable
}

with fs.open(s3_url, 'rb') as infile:
	ds = xr.open_dataset(infile, engine="grib2io", filters=filters)
	print(ds)
```
